# 7.Write a for loop to find the sum of numbers from 1 to 50

# 8.Write a for loop to find the factorial of a given number.

```python
number = int(input("enter your number :"))

factorial = 1

for i in range(1, number + 1):
    factorial = factorial * i
    print("Factorial =", factorial)
```

# 10.Write a for loop to find the largest number in a list.

```python
number = [23, 45, 56, 100, 67, 87, 98]
largest = number[0]
for i in number:
    if i > largest:
        largest = i
print("Largest number =", largest)
```
